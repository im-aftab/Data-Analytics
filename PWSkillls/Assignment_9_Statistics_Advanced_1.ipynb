1. What is a random variable in probability theory?
  - A **random variable** in probability theory is a function that assigns a numerical value to each outcome in the sample space of a random experiment, allowing uncertain outcomes to be analyzed mathematically.

2. What are the types of random variables?
  - The two main types are discrete and continuous random variables, with a third hybrid type known as mixed random variables.
    - Discrete Random Variables
      - A discrete random variable takes on a finite or countably infinite set of distinct values such as integers 0, 1, 2, 3, etc..​


      - These are used when outcomes can be counted, such as:

        - Number of heads in three coin tosses (X = 0, 1, 2, or 3)

        - Number of students present in a class
      - The probability distribution of a discrete random variable is expressed through a Probability Mass Function (PMF), which assigns a probability value to each possible outcome.​

    - Continuous Random Variables
      - A continuous random variable can take any value within a given interval or range and has infinitely many possible outcomes.
        - Examples include:​

          - Height or weight of a person

          - Time taken to finish a race
        - Probabilities for continuous random variables are described by a Probability Density Function (PDF) rather than individual probabilities, and probabilities refer to intervals of values (e.g., P(2 < X < 4)).​

    - Mixed Random Variables
      - A mixed random variable combines both discrete and continuous characteristics — part of its distribution is discrete, and part is continuous.​

      - For instance, suppose a random variable X equals −1 if a coin lands tails, and if it lands heads, X represents the value of a spinner ranging continuously between 0 and 1. Here X has both discrete (−1) and continuous (0–1) elements.


3. Explain the difference between discrete and continuous distributions.
  - Discrete Distributions
      - Discrete distributions describe the probabilities of outcomes for random variables that can take on only specific, countable values.​

      - Examples include the outcomes of rolling a die (1, 2, 3, 4, 5, 6) or the number of heads in multiple coin tosses (0, 1, 2, ..., n).​

      - Probabilities are assigned directly to each possible outcome using a Probability Mass Function (PMF).​

      - The probability that a discrete random variable equals a specific value is always nonzero (e.g., probability that X = x).​

  - Continuous Distributions
  
      - Continuous distributions apply to random variables that can take any value within a continuous range (infinite and uncountable).​

      - Examples include measurements like height, weight, or the precise time it takes to complete a task (can be any value in an interval).​

      - Probabilities are not assigned to individual values, but to intervals of values, using a Probability Density Function (PDF).​

      - The probability that a continuous random variable is exactly a particular value is always zero; probability is calculated over an interval (e.g., probability that X is between a and b).​


 4. What is a binomial distribution, and how is it used in probability?
  - The binomial distribution is a discrete probability distribution that represents the number of successes in a fixed number of independent trials, each trial having exactly two possible outcomes: success or failure

  - The binomial distribution models the probability of obtaining exactly **k** successes out of **n** trials. For example, it can tell you the probability of getting 3 heads in 5 coin tosses, the number of defective items in a batch, or passing a certain number of tests out of several attempts.


 5. What is the standard normal distribution, and why is it important?
  - The standard normal distribution (also known as the Z-distribution) is a special case of the normal distribution where the mean (μ) is 0 and the standard deviation (σ) is 1. It forms a perfectly symmetrical bell curve centered around zero, representing how values in a dataset are distributed relative to the mean.
  - Importance in Probability and Statistics
    - Simplifies calculations: Statistical tables (Z-tables) list probabilities for Z-values, eliminating the need to compute complex integrals.

    - Foundation for hypothesis testing: Z-scores help determine how extreme a sample observation is relative to a population mean, which is key in z-tests and confidence intervals.

    - Basis of the Central Limit Theorem: Many distributions approximate the standard normal when sample sizes are large, making it crucial for inferential statistics.​

    - Used for percentile and area calculations: The Z-table provides the area (probability) to the left of a specific Z value, which helps estimate probabilities of events within data ranges.

6. What is the Central Limit Theorem (CLT), and why is it critical in statistics?
  - The Central Limit Theorem (CLT) is a fundamental principle in statistics which states that, given a sufficiently large sample size, the distribution of the sample mean will be approximately normally distributed, regardless of the original population's distribution.

  - It is important because :
    - It enables inference: Even when the population distribution is unknown, statisticians can apply normal distribution techniques (like z-tests, confidence intervals) to sample means.

    - It simplifies analysis: Many probability problems become manageable because the normal distribution properties are well understood.

    - It forms the foundation of many statistical methods: The CLT is essential for hypothesis testing, estimation, and other inferential statistics.

7. What is the significance of confidence intervals in statistical analysis?
  - Significance of Confidence Intervals
    - Measure uncertainty: CIs quantify the uncertainty or variability around a sample estimate, giving a range rather than a single point estimate. This range reflects how precisely the sample data estimates the population parameter.​

    - Estimate population parameters: Since it is usually impossible to study the entire population, confidence intervals help infer the likely values of parameters based on sample data.​

    - Aid hypothesis testing: If a confidence interval excludes a null hypothesis value (e.g., zero effect), that suggests statistical significance at the chosen confidence level (e.g., 95%).​

    - Distinguish between significance and effect size: CIs show not only if an effect exists but also the range within which the effect size likely lies, which is important for practical interpretations.​

    - Communicate reliability: They provide a clear and intuitive way to communicate the reliability of statistical estimates to stakeholders or decision-makers

8. What is the concept of expected value in a probability distribution?
  - The concept of expected value in a probability distribution is the long-run average or mean value of a random variable, calculated by weighting each possible outcome by its probability of occurrence.

9. Write a Python program to generate 1000 random numbers from a normal
distribution with mean = 50 and standard deviation = 5. Compute its mean and standard deviation using NumPy, and draw a histogram to visualize the distribution.


```python
import numpy as np
import matplotlib.pyplot as plt

mean = 50
std_dev = 5
sample_size = 1000

random_numbers = np.random.normal(mean, std_dev, sample_size)

mean_value = np.mean(random_numbers)
std_dev_value = np.std(random_numbers)

plt.hist(random_numbers, bins=30, density=True, alpha=0.6, color='g')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram of Normally Distributed Random Numbers')
plt.grid(True)
plt.show()
plt.close()


print(f"\nMean: {mean_value}")
print(f"Standard Deviation: {std_dev_value}")
```

10. You are working as a data analyst for a retail company. The company has
collected daily sales data for 2 years and wants you to identify the overall sales trend.

  - daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
235, 260, 245, 250, 225, 270, 265, 255, 250, 260]

  - Explain how you would apply the Central Limit Theorem to estimate the average sales with a 95% confidence interval.

  - Write the Python code to compute the mean sales and its confidence interval.


```python
import numpy as np
import scipy.stats as stats

# Given daily sales data
daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255, 235, 260, 245, 250, 225, 270, 265, 255, 250, 260]

# Calculate sample statistics
n = len(daily_sales)
mean_sales = np.mean(daily_sales)
std_dev_sales = np.std(daily_sales, ddof=1)  # ddof=1 for sample standard deviation

# Calculate standard error
sem = std_dev_sales / np.sqrt(n)

# Find t-critical value for 95% confidence and n-1 degrees of freedom
confidence_level = 0.95
alpha = 1 - confidence_level

t_critical = stats.t.ppf(1 - alpha/2, df=n-1)

# Margin of error
margin_error = t_critical * sem

# Confidence interval
confidence_interval = (mean_sales - margin_error, mean_sales + margin_error)

# Print results
print(f"\nSample Mean Sales: {mean_sales:.2f}")
print(f"\n95% Confidence Interval: ({confidence_interval[0]:.2f}, {confidence_interval[1]:.2f})")
```
